```python
# import serial
import time
import threading

state = 2
test1 = 0
test2 = 0
lastState = None
callZynq = threading.Event()


def hydroDriver():
    
    import serial
    import time
    lastState = 0;
    lasttest1 = 0
    lasttest2 = 0
    ser = serial.Serial('COM3', 115200, timeout=1)
    ser.write(b'Hello, world!\n')
    print("Sent: Hello, world!")
    while True:
        try:
            if state!=lastState:
                lastState = state
                ser.write(f'<0,{state}>\n'.encode())
                print(f'<0,{state}>\n')
                time.sleep(0.5)
            if test1!=lasttest1:
                lasttest1 = test1
                ser.write(f'<1,{test1}>\n'.encode())
                print(f'<1,{test1}>\n')
                time.sleep(0.5)
            if test2!=lasttest2:
                lasttest2 = test2
                ser.write(f'<2,{test2}>\n'.encode())
                print(f'<2,{test2}>\n')
                time.sleep(0.5)
            if ser.in_waiting > 0:
                line = ser.readline().decode('utf-8').strip()
                if line:
                    print("Zynq Returned:", line)        
        except KeyboardInterrupt:
            print("\nStopped by user.")
            ser.close()    
        time.sleep(0.000001)
    
driver_thread = threading.Thread(target=hydroDriver, daemon=True)
driver_thread.start()

while True:
    try:
        state = int(input("Enter state: "))
        test1 = int(input("Enter test1: "))
        test2 = int(input("Enter test2: "))
    except ValueError:
        print("Invalid input.")
    time.sleep(0.01)

    
```
