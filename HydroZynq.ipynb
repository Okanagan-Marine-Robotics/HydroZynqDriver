```python
import serial
import time
import threading

state = 2
test1 = 0
test2 = 0
lastState = None
callZynq = threading.Event()


def hydroDriver():
    
    import serial
    import time
    lastState = 2;
    lasttest1 = 0
    lasttest2 = 0
    ser = serial.Serial('COM3', 115200, timeout=1)
    ser.write(b'Hello, world!\n')
    print("Sent: Hello, world!")
    while True:
        callZynq.wait()         # 💤 Wait here until event is set
        callZynq.clear()
        try:
            if state!=lastState:
                lastState = state
                ser.write(f'<0,{state}>\n'.encode())
                print(f'<0,{state}>\n')
            if test1!=lasttest1:
                lasttest1 = test1
                ser.write(f'<1,{test1}>\n'.encode())
                print(f'<1,{test1}>\n')
            if test2!=lasttest2:
                lasttest2 = test2
                ser.write(f'<2,{test2}>\n'.encode())
                print(f'<2,{test2}>\n')
            if ser.in_waiting > 0:
                line = ser.readline().decode('utf-8').strip()
                if line:
                    print("Received:", line)        
            time.sleep(0.5)
        except KeyboardInterrupt:
            print("\nStopped by user.")
            ser.close()    

driver_thread = threading.Thread(target=hydroDriver, daemon=True)
driver_thread.start()

def HydrophoneControlFunction(new_state,new_test1,new_test2):
    global state
    global test1
    global test2
    state = new_state
    callZynq.set()

while True:

    user_input = int(input("Enter state: "))
    user_input2 = int(input("Enter test1: "))
    user_input3 = int(input("Enter test2: "))
    HydrophoneControlFunction(user_input,user_input2,user_input3)
    time.sleep(0.000001)
```
