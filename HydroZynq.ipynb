```python
hydroDriver()

def hydroDriver(state=1):
    
    import serial
    import time
    
    ser = serial.Serial('COM3', 115200, timeout=1)
    ser.write(b'Hello, world!\n')
    print("Sent: Hello, world!")
    
    try:
        while True:
            ser.write(f'<0,{state}>\n'.encode())
            print(f'<0,{state}>\n')
    
            print("Waiting to receive data...")
            start_time = time.time()
            while time.time() - start_time < 1:  # read for 1 second
                if ser.in_waiting > 0:
                    line = ser.readline().decode('utf-8').strip()
                    if line:
                        print("Received:", line)
    
            time.sleep(0.5)
    except KeyboardInterrupt:
        print("\nStopped by user.")
        ser.close()
```
